      |---|---|-------|----------|----|----|---------|

      sehingga kedua ekspresi terbukti EKUIVALEN (Hukum De Morgan).

      Intuisi: "tidak benar bahwa p dan q keduanya terjadi" sama artinya dengan

      FRAME: Burung (frame umum)  
        bersayap     : ya

        bersayap      : (diwarisi: ya)

      umum berlaku mutlak untuk semua kasus turunannya.

# **`PENGEMBANGAN KEGIATAN 1 - Menghitung statistik Penjualan dan Keuntungan dengan NumPy`**

```python
import numpy as np

# Data array harga satuan, jumlah terjual, diskon (%), dan HPP per unit
harga_satuan = np.array([150000, 450000, 85000, 1200000, 250000]) # Harga dasar tiap produk
jumlah_terjual = np.array([5, 2, 10, 1, 4])                       # Jumlah unit yang dibeli
diskon_persen = np.array([0, 10, 5, 15, 0])                       # Besar persentase diskon
hpp_satuan = np.array([100000, 300000, 50000, 850000, 180000])    # Modal (HPP) per unit produk

# Hitung total penjualan sebelum diskon (Harga x Jumlah Unit)
omset_kotor = harga_satuan * jumlah_terjual

# Hitung penjualan setelah dipotong diskon
omset_bersih = omset_kotor * (1 - diskon_persen / 100)

# Hitung total modal dasar seluruh produk yang terjual
total_hpp = hpp_satuan * jumlah_terjual

# Hitung laba bersih tiap transaksi (Omset Bersih - Total Modal)
keuntungan = omset_bersih - total_hpp

# Tampilkan total omset bersih untuk masing-masing transaksi
print("Omset Bersih per TRX :", omset_bersih)

# Tampilkan keuntungan bersih untuk masing-masing transaksi
print("Keuntungan per TRX   :", keuntungan)
```

### **Penjelasan Pengembangan Kegiatan 1**

# **`PENGEMBANGAN KEGIATAN 2 - Menganalisis Penjualan Berdasarkan Kategori Produk dengan Pandas`**

```python
import pandas as pd

# 1. Memuat dataset CSV ke dalam DataFrame 'df'
url = 'https://raw.githubusercontent.com/asaputro281-boo/All-Project/refs/heads/main/ecommerce_sales_34500.csv'
df = pd.read_csv(url)

# 2. Pemeriksaan awal data
print("=== 5 BARIS PERTAMA DATASET ===")
print(df.head()) # Menampilkan 5 baris pertama data untuk melihat sampel data

print("\n=== INFORMASI STRUKTUR DATASET ===")
print(df.info()) # Menampilkan tipe data, jumlah baris/kolom, dan memori yang digunakan

print("\n=== STATISTIK DESKRIPTIF DATASET ===")
print(df.describe()) # Menampilkan ringkasan statistik (rata-rata, nilai min, max, std, dll)

# 3. Menambah kolom baru 'total_gross' hasil perkalian harga x jumlah (tanpa looping)
df["total_gross"] = df["price"] * df["quantity"]

# 4. Filter data: Mengambil baris transaksi dengan total_amount lebih dari $500
transaksi_besar = df[df["total_amount"] > 500]
print(f"\nJumlah transaksi bernilai tinggi (> $500): {len(transaksi_besar)} transaksi") # Hitung total transaksi lolos filter
print(transaksi_besar[["order_id", "category", "total_amount", "profit_margin"]].head()) # Tampilkan 5 data teratas hasil filter

# 5. Mengelompokkan data berdasarkan kategori lalu menghitung rata-rata total_amount tiap kategori
print("\nRata-rata Total Penjualan per Kategori:")
print(df.groupby("category")["total_amount"].mean())
```

### **Penjelasan Pengembangan Kegiatan 2**
Pada pengembangan Kegiatan 2 ini, Pandas dimanfaatkan untuk menganalisis dataset transaksi e-commerce berukuran besar (ecommerce_sales_34500.csv). Tahap awal dimulai dengan mengeksplorasi struktur dan statistik data menggunakan .head(), .info(), serta .describe() untuk memahami sebaran variabel finansial seperti harga, diskon, hingga jumlah barang yang terjual.  Selanjutnya, proses pengolahan data dilakukan dengan membentuk kolom baru total_gross melalui kalkulasi berbasis vektorisasi tanpa perulangan (loop). Untuk mengekstraksi wawasan bisnis, kode ini menerapkan penyaringan data (filtering) guna memisahkan transaksi bernilai tinggi (total_amount > $500). Terakhir, fitur pengelompokan (groupby) digunakan untuk menghitung rata-rata nilai transaksi pada setiap kategori produk secara otomatis, sehingga memudahkan evaluasi performa penjualan antar-kategori.

```python
# PENGEMBANGAN KEGIATAN 2: Analisis Performa & Kontribusi Kategori Produk

# 1. Mengelompokkan data berdasarkan kategori & menghitung beberapa indikator sekaligus
rekap_kategori = df.groupby("category").agg(
    total_transaksi=("order_id", "count"),   # Hitung jumlah order
    total_unit=("quantity", "sum"),          # Total unit produk terjual
    total_omset=("total_amount", "sum"),      # Total pendapatan kotor
    total_keuntungan=("profit_margin", "sum") # Total keuntungan
).reset_index()                              # Ubah kategori dari index menjadi kolom biasa

# 2. Kalkulasi rasio persentase
total_omset_keseluruhan = rekap_kategori["total_omset"].sum() # Hitung grand total omset semua kategori
rekap_kategori["kontribusi_%"] = (rekap_kategori["total_omset"] / total_omset_keseluruhan) * 100 # Persentase pangsa omset tiap kategori
rekap_kategori["profit_margin_%"] = (rekap_kategori["total_keuntungan"] / rekap_kategori["total_omset"]) * 100 # Persentase margin profit per kategori

# 3. Urutkan tabel rekapitulasi dari omset tertinggi ke terendah
rekap_kategori = rekap_kategori.sort_values(by="total_omset", ascending=False)

# Tampilkan hasil rekapitulasi dalam format tabel rapi tanpa nomor index
print("=== REKAPITULASI ANALISIS KATEGORI PRODUK ===")
print(rekap_kategori.to_string(index=False))

# 4. Filter data transaksi yang diretur saja (returned == 'Yes') lalu hitung jumlahnya per kategori
retur_per_kategori = df[df["returned"] == "Yes"].groupby("category")["order_id"].count()
print("\nJumlah Barang Diretur per Kategori:")
print(retur_per_kategori)
```

# **`PENGEMANGAN KEGIATAN 3 - Membuat Grafik Penjualan, Kategori , dan Keuntungan dengan Matplotlib (Data Dummy/Buatan)`**

```python
import matplotlib.pyplot as plt
import pandas as pd

# ==============================================================================
# 1. LOAD DATASET
# ==============================================================================
# Mengambil dataset e-commerce dari repository GitHub langsung dalam bentuk DataFrame
url = "https://raw.githubusercontent.com/asaputro281-boo/All-Project/refs/heads/main/ecommerce_sales_34500.csv"
df = pd.read_csv(url)

# ==============================================================================
# GRAFIK 1: HISTOGRAM SEBARAN NILAI TRANSAKSI
# Tujuan: Melihat distribusi/frekuensi kemunculan nilai total transaksi.
# ==============================================================================
plt.figure(figsize=(7, 4))  # Menentukan ukuran kanvas (lebar 7 inci, tinggi 4 inci)

# Membuat histogram untuk kolom 'total_amount'
# - bins=30: Membagi data ke dalam 30 kelompok rentang
# - range=(0, 1000): Membatasi visualisasi hanya untuk transaksi $0 - $1.000
plt.hist(
    df["total_amount"],
    bins=30,
    color="#1f77b4",
    edgecolor="white",
    range=(0, 1000),
)

plt.title("Sebaran Nilai Transaksi (Total Amount < $1.000)")  # Judul grafik
plt.xlabel("Total Transaksi ($)")  # Label untuk sumbu X
plt.ylabel("Jumlah Transaksi")  # Label untuk sumbu Y
plt.grid(
    axis="y", linestyle="--", alpha=0.5
)  # Garis bantu horizontal (putus-putus)
plt.show()  # Menampilkan grafik pertama

# ==============================================================================
# GRAFIK 2: DIAGRAM BATANG RATA-RATA OMSET PER KATEGORI
# Tujuan: Membandingkan performa rata-rata nilai transaksi antar kategori produk.
# ==============================================================================
# Mengelompokkan data berdasarkan 'category', menghitung rata-rata 'total_amount',
# lalu mengurutkannya dari yang tertinggi ke terendah (descending).
rata_omset = (
    df.groupby("category")["total_amount"].mean().sort_values(ascending=False)
)

plt.figure(figsize=(8, 4))  # Menentukan ukuran kanvas grafik kedua

# Membuat diagram batang tegak
# - rata_omset.index: Nama-nama kategori (sumbu X)
# - rata_omset.values: Nilai rata-rata omset (sumbu Y)
plt.bar(rata_omset.index, rata_omset.values, color="#4c9be8", edgecolor="black")

plt.title("Rata-rata Nilai Transaksi per Kategori Produk")
plt.xlabel("Kategori Produk")
plt.ylabel("Rata-rata Omset ($)")
plt.xticks(rotation=15)  # Memutar label kategori 15 derajat agar tidak bertumpuk
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()  # Menampilkan grafik kedua

# ==============================================================================
# GRAFIK 3: DIAGRAM BATANG RATA-RATA PROFIT MARGIN BERDASARKAN RENTANG HARGA
# Tujuan: Menganalisis hubungan antara kelompok harga produk dengan keuntungan (profit margin).
# ==============================================================================
# 1. Menentukan titik potong (bins) untuk mengelompokkan harga $0 - $1000 (kelipatan 100)
bins = [0, 100, 200, 300, 400, 500, 600, 700, 800, 900, 1000]

# 2. Memberikan label nama kelompok/kategori untuk tiap rentang harga
labels = [
    "$0-100",
    "$101-200",
    "$201-300",
    "$301-400",
    "$401-500",
    "$501-600",
    "$601-700",
    "$701-800",
    "$801-900",
    "$901-1000",
]

# 3. Merekategorisasi data kontinu 'price' menjadi data kategorikal rentang harga menggunakan pd.cut()
df["rentang_harga"] = pd.cut(df["price"], bins=bins, labels=labels)

# 4. Menghitung rata-rata profit margin untuk setiap kelompok rentang harga
rata_profit_margin = df.groupby("rentang_harga", observed=False)[
    "profit_margin"
].mean()

plt.figure(figsize=(9, 4))  # Menentukan ukuran kanvas grafik ketiga

# Membuat diagram batang
# .index.astype(str) digunakan untuk memastikan label rentang harga terbaca sebagai teks
plt.bar(
    rata_profit_margin.index.astype(str),
    rata_profit_margin.values,
    color="#e65100",
    edgecolor="black",
)

plt.title("Rata-rata Profit Margin Berdasarkan Rentang Harga Produk")
plt.xlabel("Rentang Harga Produk ($)")
plt.ylabel("Rata-rata Profit Margin ($)")
plt.xticks(
    rotation=30
)  # Memutar label rentang harga 30 derajat agar tulisan muat dan rapi
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()  # Menampilkan grafik ketiga
```

# **`PENGEMBANGAN KEGIATAN 4 - Mengecek Sumber, Lisensi, dan kualitas Dataset e-commerce dari Kaggle`**

```python
import matplotlib.pyplot as plt
import pandas as pd

# Memuat dataset dari Kaggle
url = "https://raw.githubusercontent.com/asaputro281-boo/All-Project/refs/heads/main/ecommerce_sales_34500.csv"
df = pd.read_csv(url)

# Perbaikan: Menggunakan nama kolom yang sesuai dengan isi CSV
kat_stats = df.groupby("category").agg(
    total_penjualan=("total_amount", "sum"), total_profit=("profit_margin", "sum")
)

# Membuat subplot side-by-side (2 grafik terpisah berdampingan)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Grafik 1: Total Penjualan
kat_stats["total_penjualan"].plot(kind="bar", color="steelblue", ax=ax1)
ax1.set_title("Total Penjualan per Kategori ($)")
ax1.set_ylabel("Penjualan ($)")
ax1.set_xlabel("Kategori Produk")
ax1.grid(axis="y", linestyle="--", alpha=0.5)
ax1.tick_params(axis="x", rotation=45)

# Grafik 2: Total Profit
colors = ["coral" if val >= 0 else "crimson" for val in kat_stats["total_profit"]]
kat_stats["total_profit"].plot(kind="bar", color=colors, ax=ax2)
ax2.set_title("Total Profit per Kategori ($)")
ax2.set_ylabel("Profit ($)")
ax2.set_xlabel("Kategori Produk")
ax2.axhline(0, color="black", linewidth=0.8)  # Garis angka 0
ax2.grid(axis="y", linestyle="--", alpha=0.5)
ax2.tick_params(axis="x", rotation=45)

plt.suptitle(
    "Analisis Performa E-Commerce per Kategori Produk",
    fontsize=14,
    fontweight="bold",
)
plt.tight_layout()
plt.show()
```

**Latihan 1**   
Dari df nilai_mahasiswa: hitung rata-rata, minimum, dan maksimum kolom
kehadiran, lalu buat histogramnya (lengkap dengan judul dan label)


```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Membaca dataset dari URL GitHub dan menyimpannya ke dalam DataFrame (df)
url = 'https://raw.githubusercontent.com/asaputro281-boo/All-Project/refs/heads/main/nilai_mahasiswa.csv'
df = pd.read_csv(url)

# 2. Menghitung statistik deskriptif khusus untuk kolom "kehadiran"
# .mean() -> mencari nilai rata-rata
rata_kehadiran = df["kehadiran"].mean()
# .min() -> mencari nilai paling rendah (minimum)
min_kehadiran = df["kehadiran"].min()
# .max() -> mencari nilai paling tinggi (maksimum)
max_kehadiran = df["kehadiran"].max()

# Menampilkan hasil statistik ke terminal/output
# :.2f berguna untuk membatasi angka di belakang koma menjadi 2 digit saja
print(f"Rata-rata Kehadiran : {rata_kehadiran:.2f}%")
print(f"Kehadiran Minimum   : {min_kehadiran}%")
print(f"Kehadiran Maksimum  : {max_kehadiran}%\n")

# 3. Membuat visualisasi data berbentuk Histogram
# Menentukan ukuran kanvas grafik (lebar 7 inci, tinggi 4 inci)
plt.figure(figsize=(7, 4))

# plt.hist() berfungsi membuat histogram dari data rentang
# bins=8 membagi data kehadiran menjadi 8 kelompok/kelompok rentang nilai
# color="skyblue" memberi warna isi batang, edgecolor="black" memberi garis tepi hitam
plt.hist(df["kehadiran"], bins=8, color="skyblue", edgecolor="black")

# Menambahkan judul grafik dan label keterangan untuk sumbu X dan Y
plt.title("Sebaran Kehadiran Mahasiswa")
plt.xlabel("Persentase Kehadiran (%)")
plt.ylabel("Jumlah Mahasiswa")

# Menambahkan garis kisi (grid) putus-putus mendatar (sumbu Y) dengan transparansi 0.7
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Menampilkan grafik yang sudah dibuat ke layar
plt.show()
```

**Latihan 2**  
Tambahkan kolom status berisi LULUS/REMIDI (batas 60 pada nilai_akhir), lalu
tampilkan berapa mahasiswa REMIDI per kelas (petunjuk: groupby + filter)

```python
import numpy as np
import pandas as pd

# Membaca kembali dataset atau memastikan df yang sudah dimuat
# Jika df sudah dimuat dari cell sebelumnya, baris ini bisa diabaikan atau dihapus.
# Namun, untuk membuat cell ini independen, kita bisa memuat ulang.
# Atau, asumsikan df sudah ada dari kegiatan 2/3 dan hanya perlu menghitung nilai_akhir.

# Memastikan kolom 'nilai_akhir' ada (hitung jika belum ada)
# Asumsi df sudah dimuat dari 'nilai_mahasiswa.csv' dan memiliki kolom 'uts' dan 'uas'
# Jika belum, uncomment baris di bawah untuk memuat df:
url = 'https://raw.githubusercontent.com/asaputro281-boo/All-Project/refs/heads/main/nilai_mahasiswa.csv'
df = pd.read_csv(url)
df["nilai_akhir"] = 0.4 * df["uts"] + 0.6 * df["uas"]

# Tambah kolom status
df['status'] = np.where(df['nilai_akhir'] >= 60, 'LULUS', 'REMIDI')

# Filter mahasiswa REMIDI dan hitung per kelas
df_remidi = df[df['status'] == 'REMIDI']
rekapan_remidi = df_remidi.groupby('kelas')['nim'].count().reset_index(name='jumlah_remidi')

print('Jumlah Mahasiswa REMIDI per Kelas:')
display(rekapan_remidi)
```

### EDA Mini Dataset Video Game Sales (vgsales.csv)

* **Sumber Dataset**: Kaggle (Video Game Sales by Gregory Smith)
* **Lisensi**: CC0: Public Domain (bebas digunakan untuk riset dan edukasi)

#### Arti 3 Kolom Utama:
1. `Platform`: Konsol atau perangkat tempat game dirilis (contoh: Wii, NES, X360, PS3).
2. `Genre`: Kategori atau jenis permainan (contoh: Action, Sports, Shooter, Role-Playing).
3. `Global_Sales`: Total penjualan kopi game secara global dalam satuan juta kopi (million copies).

### Ringkasan 3 Temuan Utama

1. **Dominasi Publisher Utama**
   * **Nintendo** memimpin sebagai *publisher* paling sukses secara global dengan total penjualan mencapai **1.786,56 juta copies**.

2. **Genre Game Paling Populer**
   * Genre game terlaris di dunia didominasi oleh **Action** (**1.751,18 juta copies**), diikuti oleh **Sports** (**1.330,93 juta copies**), dan **Shooter** (**1.037,37 juta copies**).

3. **Pangsa Pasar Berdasarkan Wilayah**
   * **Amerika Utara (NA)** menjadi pasar game terbesar dengan total penjualan **4.392,95 juta copies**, hampir dua kali lipat dari pasar **Eropa (EU)** (**2.434,13 juta copies**) dan **Jepang (JP)** (**1.291,02 juta copies**).

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Muat dataset vgsales.csv
url = 'https://raw.githubusercontent.com/asaputro281-boo/All-Project/refs/heads/main/vgsales.csv'
df_vg = pd.read_csv(url)



# -------------------------------------------------------------------
# TEMUAN 1: Top 5 Publisher dengan Penjualan Global Terbanyak
# -------------------------------------------------------------------
top_publisher = (
    df_vg.groupby("Publisher")["Global_Sales"].sum().sort_values(ascending=False).head(5)
)

print("--- TEMUAN 1: Top 5 Publisher (Juta Kopi) ---")
print(top_publisher)
print("\nAnalisis: Nintendo mendominasi pasar global dengan total penjualan 1.786,56 juta kopi.\n")

# Visualisasi Temuan 1 (Diagram Batang)
plt.figure(figsize=(8, 4))
top_publisher.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Top 5 Publisher berdasarkan Penjualan Global")
plt.xlabel("publisher")
plt.ylabel("total penjualan global (juta kopi)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# -------------------------------------------------------------------
# TEMUAN 2: 3 Genre Game Paling Laris di Dunia
# -------------------------------------------------------------------
top_genre = (
    df_vg.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False).head(3)
)

print("--- TEMUAN 2: Top 3 Genre Game Terlaris (Juta Kopi) ---")
print(top_genre)
print(
    "\nAnalisis: Genre Action dan Sports menjadi genre paling favorit di kalangan pemain game global.\n"
)

# -------------------------------------------------------------------
# TEMUAN 3: Perbandingan Total Penjualan per Wilayah Pasar
# -------------------------------------------------------------------
wilayah_sales = pd.Series(
    {
        "Amerika Utara (NA)": df_vg["NA_Sales"].sum(),
        "Eropa (EU)": df_vg["EU_Sales"].sum(),
        "Jepang (JP)": df_vg["JP_Sales"].sum(),
        "Lainnya (Other)": df_vg["Other_Sales"].sum(),
    }
)

print("--- TEMUAN 3: Total Penjualan berdasarkan Wilayah (Juta Kopi) ---")
print(wilayah_sales)
print(
    "\nAnalisis: Pasar Amerika Utara (NA) menyumbang kontribusi penjualan terbesar (4.392,95 juta kopi), hampir 2x lipat pasar Eropa."
)

# Visualisasi Temuan 3 (Diagram Batang Wilayah)
plt.figure(figsize=(7, 4))
wilayah_sales.plot(kind="bar", color=["orange", "green", "red", "purple"], edgecolor="black")
plt.title("Total Penjualan Game per Wilayah Pasar")
plt.xlabel("wilayah")
plt.ylabel("total penjualan (juta kopi)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
