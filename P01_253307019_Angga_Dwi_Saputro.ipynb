Jawaban :
Deskripsi PEAS untuk Sistem Rekomendasi Kantin Kampus:

1. Performance Measure (Ukuran Sukses):
   - Berapa banyak mahasiswa yang beneran ngebeli menu hasil rekomendasi.
   - Tingkat kepuasan atau rating dari mahasiswa setelah makan.
   - Variasi rekomendasi (menu enggak itu-itu aja tiap hari).
   - Seberapa cepat mahasiswa nemuin menu yang disukai.

2. Environment (Lingkungan):
   - Stok menu kantin yang tersedia secara real-time.
   - Profil mahasiswa (budget, alergi makanan, riwayat pesanan).
   - Jam makan (pagi, siang, atau sore).
   - Cuaca (kalau hujan, sistem cenderung nawarin makanan berkuah/hangat).

3. Actuators (Tindakan/Output):
   - Daftar pilihan menu rekomendasi yang muncul di aplikasi kantin.
   - Notifikasi promo menu harian yang dikirim ke HP mahasiswa.

4. Sensors (Input Data):
   - Riwayat pencarian dan transaksi makanan sebelumnya.
   - Input kriteria dari mahasiswa (misal: budget maksimal, opsi vegetarian).
   - Jam pemesanan dan data cuaca lokal saat itu.

Semakin banyak aturan IF - THEN ditulis , semakin cerdas sistemnya. Setujukah kamu ? Bahas dengan merujuk pelajaran era sistem pakar dan kelahir machine learning

Jawaban :
**SANGAT TIDAK SETUJU.**

1. **Pelajaran dari Era Sistem Pakar (1980-an):**
   * Dulu orang-orang mikir kalau kita nambahin puluhan ribu aturan `IF-THEN` buatan pakar, sistemnya bakal otomatis makin pintar.
   * **Kenyataannya:** Pas aturan udah kebanyakan, sistem malah makin ribet. Banyak aturan yang saling bentrok, biaya perawatannya mahal banget, dan pas nemu kasus baru sedikit aja di luar aturan, sistemnya langsung "mogok" (*brittle*). Masalah ini yang bikin AI sempat memasuki *AI Winter* kedua.

2. **Kelahiran Machine Learning (1990-an–Sekarang):**
   * Karena keterbatasan itu, pendekatan AI akhirnya bergeser. Daripada cape-cape nulis aturan satu-satu secara manual, mending **biarkan mesin yang belajar sendiri dari pola data**.
   * Kasus rumit kayak ngenalin wajah, deteksi suara, atau ngolah teks bebas itu sifatnya fleksibel banget, jadi hampir enggak mungkin kalau cuma diakalin pakai deretan aturan `IF-THEN` buatan manusia.
