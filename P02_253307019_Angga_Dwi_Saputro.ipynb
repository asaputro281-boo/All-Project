#include <stdio.h> // Header untuk mengimpor pustaka input-output dasar (seperti printf)

// Definisi fungsi untuk menghitung luas persegi panjang
int luas(int p, int l) { // Fungsi 'luas' menerima dua parameter integer (p = panjang, l = lebar) dan mengembalikan tipe int
    return p * l;        // Mengembalikan nilai hasil perkalian p dikali l ke pemanggil fungsi
}

int main() { // Fungsi utama tempat eksekusi program dimulai
    // Pemanggilan fungsi pertama
    printf("%d\n", luas(4, 6));  // Memanggil luas(4, 6) -> p=4, l=6 -> mencetak 24 lalu ganti baris

    // C tidak mendukung keyword arguments secara langsung seperti Python,
    // jadi pemanggilan tetap menggunakan urutan posisi argumen.
    printf("%d\n", luas(10, 3)); // Memanggil luas(10, 3) -> p=10, l=3 -> mencetak 30 lalu ganti baris

    return 0; // Mengembalikan nilai 0 ke sistem, menandakan program berjalan sukses
} // Penutup fungsi main

# **Versi Python Kegiatan 3**

````python
def luas(p, l):            # Membuka definisi fungsi 'luas' yang menerima 2 parameter (p = panjang, l = lebar)
    return p * l           # Mengembalikan nilai hasil perkalian p dikali l

print(luas(4, 6))          # Memanggil fungsi berdasarkan urutan posisi (p=4, l=6) lalu mencetak hasilnya (24)
print(luas(p=10, l=3))     # Memanggil fungsi dengan keyword arguments (nama parameter disebut jelas) lalu mencetak hasilnya (30)

# Output yang dihasilkan:
# 24
# 30 -- argumen boleh disebut namanya secara spesifik tanpa bergantung urutan posisi
````

````python
def konversi_nilai(angka):      # Membuka definisi fungsi 'konversi_nilai' yang menerima 1 parameter bernama 'angka'
    if angka > 82:               # Kondisi 1: jika nilai angka lebih dari 82
        return "A"              # Kembalikan huruf mutu "A" dan hentikan fungsi
    elif angka > 75:             # Kondisi 2: jika tidak > 82, tapi lebih dari 75
        return "AB"             # Kembalikan huruf mutu "AB"
    elif angka > 67:             # Kondisi 3: jika tidak > 75, tapi lebih dari 67
        return "B"              # Kembalikan huruf mutu "B"
    elif angka > 59:             # Kondisi 4: jika tidak > 67, tapi lebih dari 59
        return "BC"             # Kembalikan huruf mutu "BC"
    elif angka > 49:             # Kondisi 5: jika tidak > 59, tapi lebih dari 49
        return "C"              # Kembalikan huruf mutu "C"
    elif angka > 39:             # Kondisi 6: jika tidak > 49, tapi lebih dari 39
        return "D"              # Kembalikan huruf mutu "D"
    else:                        # Jika semua kondisi di atas tidak terpenuhi (nilai <= 39)
        return "E"              # Kembalikan huruf mutu "E"

# Uji coba pemanggilan fungsi
print(konversi_nilai(85))  # Memanggil fungsi dengan nilai 85 (masuk kondisi > 82), mencetak: A
print(konversi_nilai(70))  # Memanggil fungsi dengan nilai 70 (masuk kondisi > 67), mencetak: B
print(konversi_nilai(30))  # Memanggil fungsi dengan nilai 30 (masuk ke else), mencetak: E
````

````python
# List of Dictionaries (kumpulan data beberapa mahasiswa)
kelas = [
    {"nama": "Budi", "nilai": 85},  # Elemen ke-0: dictionary data Budi
    {"nama": "Sari", "nilai": 92},  # Elemen ke-1: dictionary data Sari
    {"nama": "Andi", "nilai": 58},  # Elemen ke-2: dictionary data Andi
    {"nama": "Rina", "nilai": 70},  # Elemen ke-3: dictionary data Rina
    {"nama": "Dedi", "nilai": 45},  # Elemen ke-4: dictionary data Dedi
]

# 1. Hitung rata-rata kelas
# List Comprehension: mengambil nilai angka dari key "nilai" di tiap dictionary 'm' dalam list 'kelas' -> [85, 92, 58, 70, 45]
semua_nilai = [m["nilai"] for m in kelas]

# Menghitung rata-rata: jumlah total nilai (sum) dibagi jumlah mahasiswa (len)
rata_rata_kelas = sum(semua_nilai) / len(semua_nilai)

# Mencetak hasil rata-rata dengan format f-string specifier ':.2f' (membatasi 2 angka di belakang koma)
print(f"Rata-rata kelas: {rata_rata_kelas:.2f}")
````
